```python
# [목적] 텍스트 분할 예제에서 사용할 원본 문서를 UTF-8 형식으로 읽어옵니다.
# data 폴더의 키워드 문서를 문자열 변수 file에 저장합니다.
# 이후 셀에서 이 문자열을 일정한 크기의 문서 조각으로 나누는 데 사용합니다.
with open("./data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read()
```

```python
# [목적] 불러온 원본 텍스트 일부를 출력하여 내용이 정상인지 확인합니다.
# 전체 문서 대신 앞 500글자만 확인해 파일 읽기와 한글 인코딩 상태를 빠르게 점검합니다.
# 확인된 file 변수는 다음 단계에서 LangChain 문서 조각으로 분할됩니다.
print(file[:500])
```

```python
# [목적] CharacterTextSplitter로 긴 문서를 문자 수 기준의 작은 문서 조각으로 나눌 준비를 합니다.
# 이 예제는 문단 구분을 우선 유지하면서, 너무 긴 텍스트를 모델이 다루기 쉬운 크기로 분할하는 방식입니다.
# 만들어진 text_splitter는 다음 셀에서 원본 텍스트를 LangChain Document 객체로 변환할 때 사용합니다.
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    # 연속된 줄바꿈을 문단 경계로 삼아 가능한 한 문단 단위로 나눕니다.
    separator="\n\n",
    # 각 조각의 최대 길이를 210자로 제한합니다.
    chunk_size=210,
    # 인접한 조각끼리 겹쳐서 보관할 글자 수이며, 여기서는 겹치지 않게 설정합니다.
    chunk_overlap=0,
    # 텍스트 길이를 셀 때 파이썬의 len 함수를 사용합니다.
    length_function=len,
)
```

```python
# [목적] 원본 텍스트를 분할한 뒤, 첫 번째 문서 조각의 크기와 내용을 확인합니다.
# create_documents는 문자열 목록을 LangChain의 Document 객체 목록으로 바꾸며, 분할된 본문은 page_content에 저장됩니다.
# 이렇게 만든 Document 목록은 이후 검색, 임베딩, 프롬프트 입력 등에 활용할 수 있습니다.
texts = text_splitter.create_documents([file])
# 첫 번째 조각의 글자 수와 실제 내용을 출력해 분할 설정이 의도대로 적용됐는지 확인합니다.
print(len(texts[0].page_content))
print(texts[0])
```

```python
# [목적] 분할된 문서에 원본 문서를 구분하는 메타데이터를 함께 저장합니다.
# 이 예제는 같은 텍스트라도 출처 정보를 붙여 Document 객체를 만드는 방식입니다.
# 메타데이터는 나중에 검색 결과의 출처를 확인하거나 특정 문서만 걸러낼 때 사용합니다.
metadatas = [
    {"document": 1},
    {"document": 2},
]

# 입력 텍스트 목록과 같은 순서의 metadatas 목록을 전달해 각 문서 조각에 출처 정보를 연결합니다.
documents = text_splitter.create_documents(
    [
        file,
        file,
    ],
    metadatas=metadatas,
)

# 첫 번째 Document 객체를 출력해 page_content와 metadata가 함께 저장됐는지 확인합니다.
print(documents[0])
```
