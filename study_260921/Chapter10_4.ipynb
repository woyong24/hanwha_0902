SemanticChunker

```python
# [목적] 의미 기반 분할 예제에서 사용할 원본 문서를 UTF-8 형식으로 읽고 내용을 확인합니다.
# data 폴더의 키워드 문서를 file 문자열에 저장한 뒤, 앞 350글자를 출력해 입력 데이터가 정상인지 점검합니다.
# 확인한 file은 다음 셀에서 문장의 의미가 달라지는 지점을 기준으로 분할하는 데 사용합니다.
with open("./data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read()

print(file[:350])
```

```python
# [목적] OpenAI 임베딩을 이용해 문맥상 의미가 바뀌는 지점에서 텍스트를 나누는 SemanticChunker를 만듭니다.
# 이 예제는 문장 길이나 줄바꿈 대신 문장들의 의미적 유사도를 비교해, 주제가 달라지는 곳을 분할 경계로 찾는 방식입니다.
# 환경 변수에서 API 키를 불러온 뒤 생성한 text_splitter는 다음 셀에서 텍스트와 Document를 만드는 데 사용합니다.
from langchain_experimental.text_splitter import SemanticChunker
from langchain_openai.embeddings import OpenAIEmbeddings
from dotenv import load_dotenv

# .env 파일의 OpenAI API 키를 환경 변수로 불러옵니다.
load_dotenv()

# OpenAIEmbeddings는 텍스트 의미를 숫자 벡터로 바꾸며, SemanticChunker가 문맥 차이를 판단하는 데 사용합니다.
text_splitter = SemanticChunker(OpenAIEmbeddings())
```

```python
# [목적] 원본 문서를 의미 단위의 텍스트 조각으로 나누고 첫 번째 결과를 확인합니다.
# split_text는 문자열 목록을 반환하며, 의미가 비슷한 문장들은 같은 조각에 유지하려고 시도합니다.
# 만들어진 chunks는 검색이나 요약 모델에 문맥이 자연스러운 입력으로 전달할 수 있습니다.
chunks = text_splitter.split_text(file)
print(chunks[0])
```

```python
# [목적] 의미 기반으로 분할한 결과를 LangChain Document 객체 목록으로 만들어 첫 내용을 확인합니다.
# create_documents는 문자열 조각을 page_content 속성에 담은 Document 객체로 변환합니다.
# Document 형식은 이후 벡터 저장소에 저장하거나 검색 체인에 연결할 때 사용하기 좋습니다.
docs = text_splitter.create_documents([file])
print(docs[0].page_content)
```
