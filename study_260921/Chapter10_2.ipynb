```python
# [목적] 재귀적 문자 분할 예제에서 사용할 원본 텍스트 문서를 UTF-8 형식으로 읽어옵니다.
# 이 예제는 문단과 줄바꿈 같은 구분자를 우선 고려해 긴 문서를 자연스러운 단위로 나누는 방식입니다.
# 읽어 온 문자열 file은 다음 단계에서 RecursiveCharacterTextSplitter에 전달할 입력값으로 사용합니다.
with open("./data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read()
```

```python
# [목적] 불러온 원본 텍스트 일부를 출력하여 분할할 데이터가 정상인지 확인합니다.
# 전체 문서 대신 앞 500글자만 확인해 파일 읽기와 한글 인코딩 상태를 빠르게 점검합니다.
# 이 확인을 마치면 같은 file 문자열을 재귀적 분할기에 전달합니다.
print(file[:500])
```

```python
# [목적] RecursiveCharacterTextSplitter로 문서 경계를 최대한 보존하며 텍스트를 나눌 설정을 만듭니다.
# 이 예제는 큰 구분자부터 순서대로 적용해, 지정한 크기를 넘지 않는 자연스러운 텍스트 조각을 만드는 방식입니다.
# 설정된 text_splitter는 다음 셀에서 원본 문자열을 LangChain Document 객체 목록으로 변환할 때 사용합니다.
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    # 각 텍스트 조각의 최대 길이를 250자로 설정합니다.
    chunk_size=250,
    # 인접한 조각이 50자씩 겹치게 해 문맥이 경계에서 끊기는 것을 줄입니다.
    chunk_overlap=50,
    # 텍스트 길이는 파이썬의 len 함수로 문자 수를 계산합니다.
    length_function=len,
    # 구분자를 정규표현식이 아닌 일반 문자열로 해석합니다.
    is_separator_regex=False,
)
```

```python
# [목적] 원본 텍스트를 LangChain Document 객체로 분할하고 앞부분의 결과를 확인합니다.
# create_documents는 문자열을 여러 조각으로 나누고, 각 조각을 page_content 속성에 담은 Document 목록으로 만듭니다.
# 이 Document 목록은 이후 임베딩 생성이나 문서 검색처럼 LangChain이 다루는 작업의 입력으로 활용할 수 있습니다.
texts = text_splitter.create_documents([file])
print(texts[0])
print("===" * 20)
print(texts[1])
```

```python
# [목적] Document 객체로 변환하지 않고, 분할된 텍스트 문자열 두 개만 간단히 확인합니다.
# split_text는 문자열 목록을 바로 반환하므로, 분할 결과 자체만 빠르게 검사할 때 적합합니다.
# 앞의 두 조각을 확인해 문단 경계와 50자 겹침 설정이 기대한 대로 적용됐는지 살펴봅니다.
text_splitter.split_text(file)[:2]
```
