```python
# 셀 목적: 현재 날짜, 프롬프트 양식, OpenAI 모델에 필요한 도구를 준비합니다.
# 핵심 개념: datetime은 컴퓨터의 현재 날짜를 가져오는 파이썬 기본 도구입니다.
from datetime import datetime
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

llm = ChatOpenAI()

# 오늘 날짜를 출력
datetime.now().strftime("%B %d")
```

```python
# 셀 목적: 실행할 때마다 오늘 날짜를 문자열로 돌려주는 함수를 만듭니다.
# 핵심 개념: 함수로 넣으면 프롬프트를 사용할 때의 날짜가 자동으로 계산됩니다.
def get_today(): 
    return datetime.now().strftime("%B %d")
```

```python
# 셀 목적: 오늘 생일인 사람을 요청하는 질문 양식을 만듭니다.
# 핵심 개념: n은 직접 입력하고, today는 get_today 함수로 자동 입력합니다.
prompt = PromptTemplate(
    template = "오늘의 날짜는 {today}입니다. 오늘이 생일인 {n}명을 나열해 주세요. 생년월일을 표기해주세요",
    input_variables=["n"],
    partial_variables={
        "today": get_today
    },
)
```

```python
# 셀 목적: 인원수를 3명으로 넣어 AI에게 전달될 질문 문장을 미리 확인합니다.
# format()은 문장만 만들며 실제 API를 호출하지 않습니다.
prompt.format(n=3)
```

```python
# 셀 목적: 질문 양식과 AI 모델을 연결하고, 3명의 답변을 실제로 요청합니다.
# 이거 과제한 날이 진짜 내 생일이었음
chain = prompt | llm
print(chain.invoke({"n": 3}).content)
print("🎉 오늘 생일인 당신도 축하합니다! 행복한 하루 보내세요! 🎂")
```

```python
# 셀 목적: 자동으로 들어가는 오늘 날짜 대신 12월 25일을 직접 넣어 결과를 비교합니다.
# 핵심 개념: 실행할 때 넘긴 today 값은 partial로 설정한 자동 날짜를 덮어쓸 수 있습니다.
print(chain.invoke({"today": "dec 25", "n": 3}).content)
```
