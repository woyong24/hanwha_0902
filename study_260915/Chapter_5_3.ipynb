```python
# 셀 목적: YAML 프롬프트, OpenAI 모델, 출력 변환과 스트리밍에 필요한 도구를 준비합니다.
# 핵심 개념: 프롬프트를 YAML에 따로 저장하면 코드와 질문 양식을 분리해 관리할 수 있습니다.
from langchain_core.prompts import load_prompt
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_teddynote.messages import stream_response

# yaml파일에서 prompt를 읽어오는 것 같음
prompt = load_prompt("test.yaml", encoding="utf-8")
prompt
```

```python
# 셀 목적: YAML에서 읽은 {fruit} 빈칸에 사과를 넣어 완성된 질문을 확인합니다.
# format()은 문장만 만들며 AI를 호출하지 않습니다.
prompt.format(fruit="사과")
```

```python
# 셀 목적: capital.yaml의 수도 질문 양식을 불러와 대한민국을 넣어 봅니다.
# encoding='utf-8'은 YAML의 한글이 깨지지 않도록 읽는 방식입니다.
prompt2 = load_prompt("capital.yaml", encoding="utf-8")
print(prompt2.format(country="대한민국"))
```

```python
# 셀 목적: 앞 셀과 같은 YAML 프롬프트 결과를 한 번 더 확인합니다.
# 기능이 중복된 연습 셀이므로 나중에는 앞 셀과 둘 중 하나만 사용해도 됩니다.
prompt2 = load_prompt("capital.yaml", encoding="utf-8")
print(prompt2.format(country="대한민국"))
```

```python
# 셀 목적: YAML 질문 양식과 AI 모델을 연결하고 답변을 실시간으로 출력합니다.
# chain.stream()은 답변을 조금씩 받고, stream_response()는 받은 내용을 화면에 보여줍니다.
# 이 셀을 실행하면 OpenAI API가 호출되어 비용이 발생할 수 있습니다.
chain = prompt2 | ChatOpenAI(model="gpt-5-nano", temperature=0) | StrOutputParser()

answer = chain.stream({"country": "대한민국"})
stream_response(answer)
```
