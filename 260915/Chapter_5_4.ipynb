```python
# 셀 목적: 나라 이름을 넣으면 수도를 묻는 질문 양식을 만듭니다.
# 핵심 개념: {country}는 나중에 실제 나라 이름으로 바뀌는 빈칸입니다.
from langchain_core.prompts import ChatPromptTemplate

# 한 줄짜리 질문 양식을 만듭니다. 아직 AI에게 질문을 보내지는 않습니다.
chat_prompt = ChatPromptTemplate.from_template("{country}의 수도는 어디인가요?")
chat_prompt
```

```python
# 셀 목적: {country} 빈칸에 대한민국을 넣어 완성된 질문을 확인합니다.
# format()은 문장만 만들며 AI를 호출하지 않으므로 비용이 발생하지 않습니다.
chat_prompt.format(country="대한민국")
```

```python
# 셀 목적: AI 대화에 필요한 역할별 메시지 양식을 만듭니다.
# system은 AI의 역할/규칙, human은 사용자 말, ai는 AI의 이전 답변 예시입니다.
chat_template = ChatPromptTemplate.from_messages(
    [
        ("system", "당신은 친절한 AI 어시스턴트입니다. 당신의 이름은 {name}입니다."),
        ("human", "반가워요!"),
        ("ai", "안녕하세요! 무엇을 도와드릴까요?"),
        ("human", "{user_input}"),
    ]
)
```

```python
# 셀 목적: 이름과 질문을 넣어 AI에 전달할 메시지 목록을 완성합니다.
# format_messages()도 메시지만 만들며 아직 AI를 호출하지 않습니다.
messages = chat_template.format_messages(
    name="Ravel", user_input="당신의 이름은 무엇입니까?"
)
messages
```

```python
# 셀 목적: OpenAI 모델을 사용할 준비를 합니다.
# ChatOpenAI()는 모델 연결 설정을 만들고, invoke()에서 실제 질문을 보냅니다.
from langchain_openai import ChatOpenAI

llm = ChatOpenAI()
llm.invoke(messages).content
```

```python
# 셀 목적: 대화 양식과 AI 모델을 하나의 실행 흐름(chain)으로 연결합니다.
# invoke()에 이름과 질문을 넣으면 양식이 완성된 뒤 AI가 답합니다. 이때 API가 호출됩니다.
chain = chat_template | llm

chain.invoke({"name": "Ravel", "user_input": "당신의 이름은 무엇입니까"}).content
```
