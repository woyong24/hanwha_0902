```python
# [목적] OpenAI의 GPT 모델로 질문에 답하고, 생성되는 내용을 실시간으로 출력하는 예제입니다.
# 모델 연결 객체를 만든 뒤 질문을 보내며, stream_response가 답변 조각을 이어서 보여 줍니다.

# 스트리밍으로 받은 답변을 보기 위한 출력 도구를 가져옵니다.
from langchain_teddynote.messages import stream_response
# LangChain에서 OpenAI GPT 모델을 사용할 수 있는 클래스를 가져옵니다.
from langchain_openai import ChatOpenAI

# 사용할 모델 이름과 답변의 일관성 정도(temperature)를 설정해 GPT 연결 객체를 만듭니다.
gpt = ChatOpenAI(
    temperature=0,
    model_name="gpt-4o"
)

# 질문의 답변을 한 번에 기다리지 않고, 생성되는 순서대로 받습니다.
answer = gpt.stream("사랑이 뭔가요?")
# 스트리밍으로 받은 답변을 노트북 화면에 출력합니다.
stream_response(answer)
```

```python
# [목적] 다른 LLM 서비스도 LangChain 객체를 통해 같은 방식으로 호출할 수 있음을 보여 주는 예제입니다.
# 모델 객체를 만든 뒤 stream()으로 질문을 보내고, 공통 출력 함수로 답변을 표시하는 흐름입니다.

# LangChain에서 DeepSeek 모델을 사용할 수 있는 클래스를 가져옵니다.
from langchain_deepseek import ChatDeepSeek

# Cohere 모델 연결 객체를 만들려는 코드입니다. 실행하려면 ChatCohere의 import와 API 키 설정이 추가로 필요합니다.
cohere = ChatCohere(temperature=0)

# Cohere 모델에 질문을 보내며 생성되는 답변을 순서대로 받습니다.
answer = cohere.stream("사랑이 뭔가요?")
# 받은 답변을 노트북 화면에 출력합니다.
stream_response(answer)
```
