1) 대화 버퍼 메모리

```python
# [목적] ConversationBufferMemory로 대화 이력을 문자열 형태로 저장하는 기본 예제
# 메모리 객체를 만든 뒤, 이후 셀에서 사용자와 AI의 대화를 차례로 누적합니다.
# ConversationBufferMemory는 이전 대화를 보관해 다음 요청에 문맥을 제공하는 LangChain 메모리입니다.
from langchain_classic.memory import ConversationBufferMemory

memory = ConversationBufferMemory()
memory
```

```python
# [목적] 기본 메모리에 첫 번째 사용자 질문과 AI 답변을 저장하는 예제
# save_context에 입력과 출력을 각각 전달하면 한 번의 대화가 메모리에 추가됩니다.
# 누적된 기록은 다음 셀에서 대화 문맥으로 조회합니다.
memory.save_context(
    inputs={
        "human": "안녕하세요, 비대면으로 은행 계좌를 개설하고 싶습니다. 어떻게 시작해야 하나요?"
    },
    outputs={
        "ai": "안녕하세요! 계좌 개설을 원하신다니 기쁩니다. 먼저, 본인 인증을 위해 신분증을 준비해 주시겠어요?"
    },
)

memory
```

```python
# [목적] 메모리에 저장된 대화 이력을 변수 형태로 조회하는 예제
# load_memory_variables는 체인이 사용할 수 있는 메모리 값을 딕셔너리로 돌려줍니다.
# 빈 딕셔너리는 이번 조회에 추가 입력값이 없음을 뜻합니다.
memory.load_memory_variables({})
```

```python
# [목적] 저장된 대화 이력만 사람이 읽기 쉬운 문자열로 출력하는 예제
# 조회 결과에서 history 키를 선택해 사용자와 AI의 누적 대화를 확인합니다.
# 메모리가 어떤 문맥을 보관했는지 점검할 때 사용합니다.
print(memory.load_memory_variables({})["history"])
```

```python
# [목적] 여러 차례의 후속 대화를 기본 메모리에 이어서 저장하는 예제
# save_context를 반복 호출해 계좌 개설 안내의 각 문답을 시간순으로 누적합니다.
# 이렇게 쌓인 기록은 모델이 앞선 진행 상황을 이어서 답하게 하는 데 필요합니다.
memory.save_context(
    inputs={"human": "네, 신분증을 준비했습니다. 이제 무엇을 해야 하나요?"},
    outputs={
        "ai": "감사합니다. 신분증 앞뒤를 명확하게 촬영하여 업로드해 주세요. 이후 본인 인증 절차를 진행하겠습니다."
    },
)

memory.save_context(
    inputs={"human": "사진을 업로드했습니다. 본인 인증은 어떻게 진행되나요?"},
    outputs={
        "ai": "업로드해 주신 사진을 확인했습니다. 이제 휴대폰을 통한 본인 인증을 진행해 주세요. 문자로 발송된 인증번호를 입력해 주시면 됩니다."
    },
)

memory.save_context(
    inputs={"human": "인증번호를 입력했습니다. 계좌 개설은 이제 어떻게 하나요?"},
    outputs={
        "ai": "본인 인증이 완료되었습니다. 이제 원하시는 계좌 종류를 선택하고 필요한 정보를 입력해 주세요. 예금 종류, 통화 종류 등을 선택할 수 있습니다."
    },
)
```

```python
# [목적] 여러 대화가 추가된 뒤 전체 문자열 이력을 확인하는 예제
# history 값을 출력해 앞서 저장한 문답이 순서대로 누적되었는지 확인합니다.
# 기본 설정에서는 대화가 하나의 긴 텍스트로 반환됩니다.
print(memory.load_memory_variables({})["history"])
```

```python
# [목적] 계좌 개설의 마지막 문답을 추가해 전체 대화 흐름을 완성하는 예제
# 두 번의 save_context 호출로 안내 절차와 완료 응답을 기존 메모리에 계속 저장합니다.
# 이후 출력 셀에서 완성된 이력이 올바르게 연결되는지 확인합니다.
memory.save_context(
    inputs={"human": "정보를 모두 입력했습니다. 다음 단계는 무엇인가요?"},
    outputs={
        "ai": "입력해 주신 정보를 확인했습니다. 계좌 개설 절차가 거의 끝났습니다. 마지막으로 이용 약관에 동의해 주시고, 계좌 개설을 최종 확인해 주세요."
    },
)
memory.save_context(
    inputs={"human": "모든 절차를 완료했습니다. 계좌가 개설된 건가요?"},
    outputs={
        "ai": "네, 계좌 개설이 완료되었습니다. 고객님의 계좌 번호와 관련 정보는 등록하신 이메일로 발송되었습니다. 추가적인 도움이 필요하시면 언제든지 문의해 주세요. 감사합니다!"
    },
)
```

```python
# [목적] 기본 메모리에 저장된 최종 대화 이력을 출력하는 예제
# history 키의 값을 출력하여 처음부터 마지막까지 누적된 문답을 한 번에 봅니다.
# 이 결과는 문자열 기반 메모리의 저장 형태를 확인하는 데 사용합니다.
print(memory.load_memory_variables({})["history"])
```

```python
# [목적] return_messages 옵션으로 메시지 객체 형태의 대화 이력을 저장하는 예제
# 앞선 예제와 같은 대화를 저장하되, 조회 결과를 문자열 대신 메시지 목록으로 받습니다.
# 메시지 객체가 필요한 채팅 모델이나 체인에 전달할 때 사용하는 방식입니다.
memory = ConversationBufferMemory(return_messages=True)

memory.save_context(
    inputs={
        "human": "안녕하세요, 비대면으로 은행 계좌를 개설하고 싶습니다. 어떻게 시작해야 하나요?"
    },
    outputs={
        "ai": "안녕하세요! 계좌 개설을 원하신다니 기쁩니다. 먼저, 본인 인증을 위해 신분증을 준비해 주시겠어요?"
    },
)

memory.save_context(
    inputs={"human": "네, 신분증을 준비했습니다. 이제 무엇을 해야 하나요?"},
    outputs={
        "ai": "감사합니다. 신분증 앞뒤를 명확하게 촬영하여 업로드해 주세요. 이후 본인 인증 절차를 진행하겠습니다."
    },
)

memory.save_context(
    inputs={"human": "사진을 업로드했습니다. 본인 인증은 어떻게 진행되나요?"},
    outputs={
        "ai": "업로드해 주신 사진을 확인했습니다. 이제 휴대폰을 통한 본인 인증을 진행해 주세요. 문자로 발송된 인증번호를 입력해 주시면 됩니다."
    },
)
```

```python
# [목적] 메시지 객체 목록으로 반환된 대화 이력을 확인하는 예제
# return_messages=True 설정 때문에 history에는 HumanMessage와 AIMessage 객체가 순서대로 들어 있습니다.
# 이 형식은 역할별 메시지 구조를 유지한 채 후속 채팅 모델에 전달할 수 있습니다.
memory.load_memory_variables({})["history"]
```
