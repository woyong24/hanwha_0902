3) HWP 로더

```python
# [목적] HWPLoader로 한글(HWP) 문서를 LangChain 문서 목록으로 읽어 오는 예제
# HWPLoader는 HWP 파일의 본문을 LangChain Document 객체로 변환하는 전용 로더이며, 문서 검색과 분할의 입력으로 사용할 수 있습니다.
# 파일 경로로 로더를 만든 뒤 load()를 호출해 변환된 문서를 docs 변수에 저장합니다.
from langchain_teddynote.document_loaders import HWPLoader

# HWP Loader 객체 생성
loader = HWPLoader("../data/디지털 정부혁신 추진계획.hwp")

docs = loader.load()  # 문서 로드
```

```python
# [목적] 불러온 HWP 문서의 본문이 정상적으로 추출됐는지 확인하는 예제
# docs[0]은 첫 번째 Document이며, page_content에는 문서에서 추출한 텍스트 본문이 들어 있습니다.
# 앞 1,000자만 출력해 긴 전체 문서를 열지 않고 내용과 추출 상태를 빠르게 점검합니다.
# 결과 출력
print(docs[0].page_content[:1000])
```
