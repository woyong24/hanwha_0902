2) PDF 로더

```python
# [목적] PDF 파일 경로를 설정하고 환경 변수를 불러오는 예제
# load_dotenv()는 .env 파일의 설정값을 프로그램에서 사용할 수 있게 불러오며, 비밀 키가 필요한 다음 단계에 대비합니다.
# PDF 경로는 FILE_PATH 변수에 저장해 로더를 만들 때 같은 파일을 일관되게 지정합니다.
from dotenv import load_dotenv

load_dotenv()

FILE_PATH = "../data/SPRI_AI_Brief_2023년12월호_F.pdf"
```

```python
# [목적] PyPDFLoader로 PDF 본문을 읽고 일부 내용을 확인하는 예제
# PyPDFLoader는 PDF의 페이지별 내용을 LangChain Document 객체 목록으로 바꾸며, load()가 실제 읽기 작업을 수행합니다.
# 읽은 결과에서 열한 번째 페이지의 앞 300자를 출력해 텍스트 추출이 정상인지 확인합니다.
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(FILE_PATH)  # 파일 경로 설정

docs = loader.load()  # PDF 로더 초기화

print(docs[10].page_content[:300])  # 문서의 내용 출력
```
