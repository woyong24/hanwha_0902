```python
# [목적] AI 모델을 사용하기 위한 도구와 API 키 환경을 준비합니다.
# [복습] import는 기능을 가져오고 load_dotenv는 .env 설정을 읽습니다.
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field

load_dotenv()
```

```python
# [목적] 모델이 반환할 이메일 요약의 모양을 미리 정합니다.
# [복습] 필드를 정해두면 AI 결과를 일정한 형식으로 사용할 수 있습니다.
# with_structured_output()이 이 구조에 맞는 Python 객체를 바로 반환합니다.
class EmailSummary(BaseModel):
    person: str = Field(description="메일을 보낸 사람")
    email: str = Field(description="메일을 보낸 사람의 이메일 주소")
    subject: str = Field(description="메일 제목")
    summary: str = Field(description="메일 본문을 요약한 텍스트")
    date: str = Field(description="메일 본문에 언급된 미팅 날짜와 시간")
```

```python
# [목적] AI에게 요약을 맡길 원본 이메일 내용입니다.
# [활용] 실제 프로그램에서는 파일이나 메일 API의 내용을 넣습니다.
email_conversation = """From: 김철수 (chulsoo.kim@bikecorporation.me)
To: 이은채 (eunchae@teddyinternational.me)
Subject: "ZENESIS" 자전거 유통 협력 및 미팅 일정 제안

안녕하세요, 이은채 대리님,

저는 바이크코퍼레이션의 김철수 상무입니다. 최근 보도자료를 통해 귀사의 신규 자전거 "ZENESIS"에 대해 알게 되었습니다.
ZENESIS 모델의 기술 사양, 배터리 성능, 디자인 정보가 담긴 상세 브로슈어를 요청드립니다.
또한 협력 가능성을 논의하기 위해 다음 주 화요일(1월 15일) 오전 10시에 귀사 사무실에서 미팅을 제안합니다.

감사합니다.

김철수
상무이사
바이크코퍼레이션
"""
```

```python
# [목적] AI 모델이 정상적으로 응답하는지 간단히 시험합니다.
# [복습] invoke는 질문을 보내고 답변 전체를 한 번에 받습니다.
llm = ChatOpenAI(
    temperature=0, model_name="gpt-5-nano"
)

llm.invoke("대한민국의 수도는 어디인가요?")
```

```python
# [목적] 모델이 EmailSummary 형식으로 바로 답하도록 설정합니다.
# [핵심] 별도 파서 없이 정해둔 필드의 객체를 반환합니다.
llm_with_structured = ChatOpenAI(
    temperature=0, model_name="gpt-5-nano"
).with_structured_output(EmailSummary)
```

```python
# [목적] 이메일을 보내고 EmailSummary 객체로 요약 결과를 받습니다.
# [활용] answer.person처럼 필요한 정보만 골라 사용할 수 있습니다.
answer = llm_with_structured.invoke(email_conversation)

answer.person
```
